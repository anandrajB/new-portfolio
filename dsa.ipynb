```python
""" 
1. Two Sum 
Input: nums = [2,7,11,15], target = 9
Output: [0,1]
Explanation: Because nums[0] + nums[1] == 9, we return [0, 1].
"""

numbers = [2,7,11,15]

target = 9 

""" brute force approach """
# founded_index = None
# for i in range(len(numbers)):
#     for j in range(i+1 ,len(numbers)):
#         if numbers[i] + numbers[j] == target:
#             founded_index = [i,j]
#         # print("the index" , numbers[i] ,numbers[j])

# print("the data" , founded_index)



"""dict baased approach"""
seen = {}
for index , data in enumerate(numbers):
    # print("the idnex" , index , data )
    complement_data = target - data 
    if complement_data in seen:
        [seen[complement_data] , index]
    else:
         # print("the complimen" , complement_data)
        seen[data] = index 

print(seen)

```

```python
"""
Diagonal sum

Input: mat = [[1,2,3],
              [4,5,6],
              [7,8,9]]
Output: 25


"""

mat = [[1,2,3],
    [4,5,6],
    [7,8,9]]

diagSum = 0
n = len(mat)
for i in range(len(mat)):
    diagSum += mat[i][i]

for i in range(len(mat)):
    diagSum += mat[i][n-1-i]


if n%2:
    diagSum -= mat[n//2][n//2]
    # diagSum -= mat[n//2]
print(diagSum)
```

```python
""" maximum product subarray """

nums = [2,3,-2,4]

max_prod = nums[0]

for i in range(len(nums)):
    mul = 1 
    for j in range(i , len(nums)):
        mul *= nums[j]
        max_prod = max(max_prod , mul)

print(max_prod)
```
