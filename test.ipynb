```python
from concurrent.futures import ThreadPoolExecutor
import time
import threading


def fetch_data(url):
    time.sleep(1)  # Simulate network request
    return f"Data from {url}"


urls = ["https://site1.com", "https://site2.com", "https://site3.com"]

# max_workers sets the maximum number of threads in the pool
with ThreadPoolExecutor(max_workers=3) as executor:
    # executor.map applies the function to each item concurrently
    results = executor.map(fetch_data, urls)

for result in results:
    print(result)
```

```python
import multiprocessing as mp
from concurrent.futures import ProcessPoolExecutor


def compute_range(start, end):
    return sum(range(start, end))


if __name__ == "__main__":
    ctx = mp.get_context("fork")
    with ProcessPoolExecutor(max_workers=3, mp_context=ctx) as executor:
        results = list(
            executor.map(
                compute_range,
                [1, 1000000, 2000000],
                [1000000, 2000000, 3000000],
            )
        )

    for result in results:
        print(result)
```

```python
class ParentClass:
    def __init__(self):
        self.public_data = "Hello"

    def __some_func(self):
        """This is a protected method."""
        print("Executing protected logic inside the Parent class.")

    def public_method(self):
        """Public methods can call protected methods safely."""
        print("Calling from a public method:")
        self.__some_func()


class ChildClass(ParentClass):
    def test_protected(self):
        # Child classes can freely access the parent's protected method
        print("Calling parent's protected method from Child:")
        self.__some_func()


# --- How they behave in practice ---

obj = ParentClass()

# 1. Standard Usage (Inside the class)
# obj.public_method()

# # 2. Inheritance Usage (Inside the child class)
# child_obj = ChildClass()
# child_obj._some_func()  # This is allowed and expected

# # 3. External Usage (Technically possible, but strongly discouraged)
# # Your IDE (like VS Code or PyCharm) will likely flag this with a warning hint.
obj._ParentClass__some_func()  # This will raise an AttributeError because it's name-mangled
```

```python
class Student:
    school = "ABC School"

    def display(self, name):
        print(f"Student Name: {name}")

    @classmethod
    def change_school(cls, new_school):
        cls.school = new_school

    @classmethod
    def show_school(cls):
        print(f"School: {cls.school}")

    @staticmethod
    def is_adult(age):
        return age >= 18


s = Student()
s.display("Kevin")
Student.change_school("XYZ School")
print(Student.is_adult(20))
print(Student.show_school())
```

```python
import concurrent.futures
import threading
import time


def my_func(task_id):
    for i in range(2):
        # Print the exact worker thread handling this task
        print(
            f"Worker [{threading.current_thread().name}] handling Task {task_id} (iteration {i})"
        )
        time.sleep(0.5)


if __name__ == "__main__":
    # Create a pool with exactly 6 worker threads
    with concurrent.futures.ThreadPoolExecutor(max_workers=2) as executor:
        # Submit 4 tasks to the pool of 6 workers
        for task_id in range(4):
            executor.submit(my_func, task_id)
```

```python
@dataclass
class Food:
    name: str
    origin: str
    food_rating: float
    calories: int

    def __repr__(self):
        return "Food Info : {} Rating : {} , calories {}".format(
            self.name, self.food_rating, self.calories
        )

    def __post_init__(self):
        if not (0 <= self.food_rating <= 5):
            raise ValueError("Food rating must be between 0 and 5.")

    @property
    def get_food_info(self):
        return "{} {}".format(self.origin, self.calories)


@dataclass(slots=True)
class FoodOption:

    food: Food
    quantity: int

    def get_total_calories(self):
        return self.food.calories * self.quantity


f = [
    FoodOption(Food("Pizza", "Italy", 4.5, 300), 2),
    FoodOption(Food("Burger", "USA", 4.0, 500), 1),
    FoodOption(Food("Sushi", "Japan", 5.0, 200), 3),
]

ordered_food = sorted(f, key=lambda x: x.food.food_rating, reverse=True)


cleaned_dict = [
    {
        food.food.name: {
            **food.food.__dict__,
            "calories": food.food.calories * food.quantity,
            "quantity": food.quantity,
        }
    }
    for food in ordered_food
]

print(cleaned_dict)
```

```python
from functools import wraps


def customdecorator(func):

    @wraps(func)
    def wrapper(*args , **kwargs):
        result = func(*args , **kwargs)
        return result + "world"

    return wrapper

@customdecorator
def hello():
    return "Hello"


print(hello())
```

```python

from functools import wraps
# basic cache decorator class 

from dataclasses import dataclass , field
from typing import Any

@dataclass
class LruCache :


    cache_data : dict = field(default_factory= dict )

    def __call__(self ,func):
        @wraps(func)
        def wrapper(*args , **kwargs):
            cache_key = [f"{k}_{v}" for k, v in kwargs.items()]
            if cache_key[0] in self.cache_data:
                return self.cache_data
            result = func(*args , **kwargs)
            self.cache_data[cache_key[0]] = result
           
            return result
        return wrapper




@LruCache()
def somefunction(x):
    return x 


print(somefunction(x = "8yquw8wqhquhwiuqwiuqhuwihquwhuqwhqhwoiuqiow"))
```

```python
from collections import defaultdict , Counter
distribution = {}
die_a = [1,2,3,4,5,6]
die_b = [1,2,3,4,5,6]

total = len(die_a) * len(die_b)
distribution = Counter()

for a in die_a:
    for b in die_b:
        distribution[a + b] += 1 


for s in sorted(distribution):
    print(
        f"Sum = {s}, Count = {distribution[s]}, Probability = {distribution[s]/total}"
    )
```

```python
class Solution:
    def isValid(self, s: str) -> bool:
        while "()" in s or "[]" in s or "{}" in s:
            s = s.replace("()", "")
            s = s.replace("[]", "")
            s = s.replace("{}", "")
        return s == ""


obj = Solution()
s = "{()}"
ds = obj.isValid(s)

print("the " , ds)
```
